```python


# Library for data manipulation
import pandas as pd

# Library for numerical operations
import numpy as np

# Libraries for data visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Libraries for machine learning
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Ignore warning messages
import warnings
warnings.filterwarnings("ignore")
```

```python
# Read the dataset from the CSV file
df = pd.read_csv(r"C:\Users\Admin\Desktop\DATA_Mining\loan_data.csv")

# Display the first five records
df.head()
```

```python
# Display the number of rows and columns
print("Dataset Shape:")
print(df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns)

# Display information about the dataset
print("\nDataset Information:")
df.info()
```

```python
# Display summary statistics for numerical columns
df.describe()
```

```python
# Count missing values in each column
print(df.isnull().sum())
```

```python
# Select all categorical columns
categorical_columns = df.select_dtypes(include="object").columns

# Fill missing values with the most frequent value (mode)
for column in categorical_columns:
    df[column].fillna(df[column].mode()[0], inplace=True)
```

```python
# Select all numerical columns
numerical_columns = df.select_dtypes(include=["int64", "float64"]).columns

# Fill missing values with the median value
for column in numerical_columns:
    df[column].fillna(df[column].median(), inplace=True)
```

```python
# Count duplicate rows
print("Duplicate Records:", df.duplicated().sum())
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Verify duplicates are removed
print("Duplicate Records After Removal:", df.duplicated().sum())
```

```python
# Create an object of LabelEncoder
label_encoder = LabelEncoder()

# Convert all categorical columns into numeric values
for column in categorical_columns:
    df[column] = label_encoder.fit_transform(df[column])
```

```python
# Separate the input features and the target variable

# X contains all the input columns except Loan_Status
X = df.drop("Loan_Status", axis=1)

# y contains the output column (Loan Approval Status)
y = df["Loan_Status"]

# Display the shape of input and output data
print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
# Split the dataset into training data and testing data

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Display the size of training and testing data
print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
# Create the Decision Tree model
decision_tree = DecisionTreeClassifier(random_state=42)

# Train the model using the training data
decision_tree.fit(X_train, y_train)
```

```python
# Predict loan status for the test data

predicted_status = decision_tree.predict(X_test)

# Display the first 10 predictions
print(predicted_status[:10])
```

```python
# Calculate the accuracy of the model

accuracy = accuracy_score(y_test, predicted_status)

print("Model Accuracy:", round(accuracy * 100, 2), "%")
```

```python
# Display the confusion matrix

conf_matrix = confusion_matrix(y_test, predicted_status)

print("Confusion Matrix")
print(conf_matrix)
```

```python
# Display Precision, Recall and F1-Score

report = classification_report(y_test, predicted_status)

print(report)
```
