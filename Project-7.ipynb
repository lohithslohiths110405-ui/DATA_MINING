```python
# Import libraries for data handling
import pandas as pd
import numpy as np

# Import libraries for visualization
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")
```

```python
# Load the healthcare dataset
df = pd.read_csv(r"C:\Users\Admin\Desktop\DATA_Mining\healthcare_dataset.csv")

# Display the first five rows
df.head()
```

```python
# Display number of rows and columns
print("Dataset Shape:")
print(df.shape)
```

```python
# Display column names
print("\nColumn Names:")
print(df.columns)
```

```python

# Display information about the dataset
print("\nDataset Information:")
df.info()
```

```python
# Display data types of all columns
print(df.dtypes)
```

```python
# Display statistical summary of numerical columns
df.describe()
```

```python
# Check missing values in each column
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Check duplicate records
print("Duplicate Records:", df.duplicated().sum())

# Remove duplicate records
df = df.drop_duplicates()

# Verify duplicates are removed
print("Duplicate Records After Removal:", df.duplicated().sum())
```

```python
# Convert admission date to datetime format
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])

# Convert discharge date to datetime format
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"])
```

```python
# Check if any missing values remain
print(df.isnull().sum())
```

```python
# Display the first five rows of the dataset
df.head()
```

```python
# Count the number of male and female patients
print(df["Gender"].value_counts())

# Create a bar chart
plt.figure(figsize=(6,4))
sns.countplot(x="Gender", data=df)

plt.title("Number of Patients by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")

plt.show()
```

```python
# Display the age distribution of patients

plt.figure(figsize=(8,5))

plt.hist(df["Age"], bins=10)

plt.title("Age Distribution of Patients")
plt.xlabel("Age")
plt.ylabel("Number of Patients")

plt.show()
```

```python
# Display different blood groups

plt.figure(figsize=(8,5))

sns.countplot(x="Blood Type", data=df)

plt.title("Blood Group Distribution")
plt.xlabel("Blood Type")
plt.ylabel("Number of Patients")

plt.show()
```

```python
# Display the number of patients for each medical condition

plt.figure(figsize=(10,5))

sns.countplot(x="Medical Condition", data=df)

plt.title("Patients by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Number of Patients")

plt.xticks(rotation=45)

plt.show()
```

```python
# Display the number of admissions based on admission type

plt.figure(figsize=(6,4))

sns.countplot(x="Admission Type", data=df)

plt.title("Admission Type Distribution")
plt.xlabel("Admission Type")
plt.ylabel("Number of Patients")

plt.show()
```

```python
# Display patient test results

plt.figure(figsize=(6,4))

sns.countplot(x="Test Results", data=df)

plt.title("Test Results")
plt.xlabel("Result")
plt.ylabel("Number of Patients")

plt.show()
```

```python
# Display the hospitals with the highest number of patients

top_hospitals = df["Hospital"].value_counts().head(10)

plt.figure(figsize=(10,5))

top_hospitals.plot(kind="bar")

plt.title("Top 10 Hospitals")
plt.xlabel("Hospital")
plt.ylabel("Number of Patients")

plt.xticks(rotation=45)

plt.show()
```

```python
# Display the insurance providers with the highest number of patients

top_insurance = df["Insurance Provider"].value_counts()

plt.figure(figsize=(8,5))

top_insurance.plot(kind="bar")

plt.title("Insurance Provider Distribution")
plt.xlabel("Insurance Provider")
plt.ylabel("Number of Patients")

plt.xticks(rotation=45)

plt.show()
```

```python
# Display the billing amount distribution

plt.figure(figsize=(8,5))

plt.hist(df["Billing Amount"], bins=20)

plt.title("Billing Amount Distribution")
plt.xlabel("Billing Amount")
plt.ylabel("Number of Patients")

plt.show()
```

```python
# Calculate the average billing amount for each medical condition

average_bill = df.groupby("Medical Condition")["Billing Amount"].mean()

print(average_bill)

# Display the result as a bar chart

plt.figure(figsize=(10,5))

average_bill.plot(kind="bar")

plt.title("Average Billing Amount by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Average Billing Amount")

plt.xticks(rotation=45)

plt.show()
```

```python
# Select only numerical columns
numeric_data = df.select_dtypes(include=["int64", "float64"])

# Display the correlation matrix

plt.figure(figsize=(8,6))

sns.heatmap(numeric_data.corr(),
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```
